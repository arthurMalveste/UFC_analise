```python
df["TIME_original"] = df["TIME"]
```

```python
partes = df["TIME"].str.split(":", expand=True).astype(int)
df["TIME"] = partes[0] * 60 + partes[1]
```

```python
df['EVENT'] = df['EVENT'].str.strip()
```

```python
df['DETAILS'] = df['DETAILS'].str.strip()
```

```python
df['METHOD'] = df['METHOD'].str.strip()
```

```python
# df['EVENT'].unique()
df['METHOD'].unique()
```

```python
formato_novo = df.count() - df[(df['TIME FORMAT'] == '5 Rnd (5-5-5-5-5)') | (df['TIME FORMAT'] == '3 Rnd (5-5-5)')].count()

print(f"Total de lutas: {df.count()}")
print(f"Total de lutas com formato diferente de 5 Rnd (5-5-5-5-5) ou 3 Rnd (5-5-5): {formato_novo}")
```

```python
df = df[df['TIME FORMAT'].isin(['5 Rnd (5-5-5-5-5)', '3 Rnd (5-5-5)'])]
```

```python
df = df[df["BOUT"].notna()]
```

```python
df.info()
```

```python
# 1) Identificar: a URL é o ID da luta, então tem que ser única
print("URL é única?", df["URL"].is_unique)
print("Linhas com URL repetida:", df["URL"].duplicated().sum())

# 2) Olhar as duplicatas lado a lado antes de apagar qualquer coisa
duplicadas = df[df["URL"].duplicated(keep=False)].sort_values("URL")
duplicadas[["EVENT", "BOUT", "OUTCOME", "METHOD", "ROUND", "TIME"]]
```

```python
# 3) Corrigir: manter a versão cujo nome de evento existe na tabela oficial de eventos
eventos = pd.read_csv("../dados/brutos/ufc_event_details.csv")
eventos_validos = set(eventos["EVENT"].str.strip())

df["evento_valido"] = df["EVENT"].isin(eventos_validos)

df = (
    df.sort_values("evento_valido", ascending=False)  
      .drop_duplicates(subset="URL", keep="first")    
      .drop(columns="evento_valido")
      .sort_index()                                 
)

# 4) Conferir
print("URL é única agora?", df["URL"].is_unique)
print("Total de linhas:", len(df))  
```
