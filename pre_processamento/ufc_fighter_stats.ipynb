O pré-processamento típico segue essa ordem:

- Olhar os tipos (df.info()) e ver o que devia ser número/data mas veio como object/str

- Olhar valores únicos (df["COLUNA"].unique()) pra achar sujeira: espaços, formatos inconsistentes.

- Limpar (strip, trocar separador, etc.)

- Converter pro tipo certo

- Só então calcular estatísticas, agrupar, plotar

```python
import pandas as pd

df = pd.read_csv("../dados/brutos/ufc_fighter_stats.csv")
```

```python
df.info()
```

```python
df = df.replace("--", None)
```

HEIGHT: pés e polegadas para cm

```python
partes = df["HEIGHT"].str.replace('"', "").str.split("' ", expand=True).apply(pd.to_numeric)
df["HEIGHT_CM"] = (partes[0] * 12 + partes[1]) * 2.54
```

REACH: polegadas para cm

```python
df["REACH_CM"] = pd.to_numeric(df["REACH"].str.replace('"', "")) * 2.54
```

WEIGHT: libras para kg

```python
df["WEIGHT_KG"] = pd.to_numeric(df["WEIGHT"].str.replace(" lbs.", "")) * 0.4536
```

DOB: texto para data

```python
df["DOB"] = pd.to_datetime(df["DOB"], format="%b %d, %Y")
```

Remover as colunas de texto originais



```python
df = df.drop(columns=["HEIGHT", "REACH", "WEIGHT"])
```

```python
df.info()
df[['HEIGHT_CM', 'REACH_CM', 'WEIGHT_KG']].describe()
```

salvando

```python
df.to_csv("../dados/processados/ufc_fighter_stats.csv", index=False)
```
